## Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowballsStemmer=SnowballStemmer('english')
```

```python
for word in words:
    print(word+"--->"+snowballsStemmer.stem(word))
```

```python
porter = PorterStemmer()
snowball = SnowballStemmer("english")

# Removes common endings using a regular expression
regexp = RegexpStemmer(
    r"ing$|ed$|es$|s$|ly$|ment$|ness$|ation$|er$",
    min=4
)

words = [
    "eat", "eats", "eating", "eaten",
    "write", "writes", "writing", "writer", "written",
    "program", "programs", "programming", "programmed", "programmer",
    "study", "studies", "studying", "studied",
    "play", "plays", "playing", "played", "player",
    "run", "runs", "running", "runner",
    "connect", "connected", "connecting", "connection", "connections",
    "organize", "organized", "organizing", "organization",
    "develop", "developed", "developing", "developer", "development",
    "compute", "computed", "computing", "computer", "computation",
    "differ", "different", "difference", "differently",
    "final", "finally", "finalized", "finalization",
    "history", "historical", "historically",
    "happy", "happily", "happiness",
    "care", "caring", "careful", "carefully", "careless",
    "use", "uses", "using", "useful", "useless",
    "nation", "national", "nationality", "international",
    "create", "created", "creating", "creation", "creative",
    "classify", "classified", "classification", "classifier",
    "predict", "predicted", "prediction", "predictive",
    "normalize", "normalized", "normalization",
    "token", "tokens", "tokenize", "tokenized", "tokenization"
]

print(f"{'Word':20} {'Porter':15} {'Snowball':15} {'Regexp':15}")
print("-" * 70)

for word in words:
    porter_result = porter.stem(word)
    snowball_result = snowball.stem(word)
    regexp_result = regexp.stem(word)

    print(
        f"{word:20} "
        f"{porter_result:15} "
        f"{snowball_result:15} "
        f"{regexp_result:15}"
    )
```

```python
import pandas as pd

results = []

for word in words:
    results.append({
        "Original Word": word,
        "Porter Stemmer": porter.stem(word),
        "Snowball Stemmer": snowball.stem(word),
        "Regexp Stemmer": regexp.stem(word)
    })

df = pd.DataFrame(results)

df
```

```python
different_results = df[
    (df["Porter Stemmer"] != df["Snowball Stemmer"]) |
    (df["Porter Stemmer"] != df["Regexp Stemmer"]) |
    (df["Snowball Stemmer"] != df["Regexp Stemmer"])
]

different_results
```

 # Wordnet Lemmatizer

```python
import nltk

nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
'''
POS- Noun-n
verb-v
adjective-a
adverb-r
'''

lemmatizer.lemmatize("going", pos="v")
```

```python
words = [
    "going",
    "eating",
    "eaten",
    "running",
    "written",
    "studies",
    "children",
    "mice",
    "better"
]
for word in words:
    print(
        f"{word:12} "
        f"noun: {lemmatizer.lemmatize(word, pos='n'):12} "
        f"verb: {lemmatizer.lemmatize(word, pos='v'):12} "
        f"adjective: {lemmatizer.lemmatize(word, pos='a')}"
    )
```
