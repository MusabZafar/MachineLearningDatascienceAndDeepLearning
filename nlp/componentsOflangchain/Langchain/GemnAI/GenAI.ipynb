#### Getting started With Langchain And Google AI

In this quickstart we'll see how to:

- Get setup with LangChain, LangSmith and LangServe
- Use the most basic and common components of LangChain: prompt templates, models, and output parsers.
- Build a simple application with LangChain
- Trace your application with LangSmith
- Serve your application with LangServe

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['GOOGLE_API_KEY']=os.getenv('GOOGLE_API_KEY')
## langsmith tracking
os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"]="true"
os.environ["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
```

```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
load_dotenv()

# Use an active, supported model version
llm=ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite", 
    temperature=0.7, 
    timeout=15
)
```

```python
import socket

try:
    ip = socket.gethostbyname("googleapis.com")
    print(f"Connected! Resolved IP: {ip}")
except Exception as e:
    print(f"DNS resolution failed: {e}")
```

```python
import httpx

try:
    response = httpx.get("https://generativelanguage.googleapis.com", timeout=10)
    print(f"Connection successful! Status code: {response.status_code}")
except Exception as e:
    print(f"Connection failed: {e}")
```

```python
print("Invoking model...")
## Input and get response from LLM
response = llm.invoke("Hello, confirm you are working.")
print("\nResponse received:")
print(response.content)
```

```python
### Chatprompt Template
from langchain_core.prompts import ChatPromptTemplate
prompt=ChatPromptTemplate(
    [
        ("system","You are an expert AI Engineer. Provide me answers on the question"),
        ("user","{input}")
    ]
)
prompt
```

```python
chain=prompt|llm
response=chain.invoke({"input":"Can you tell me about AI"})
print(response)
```

```python
type(response)
```

```python
## stroutput parser
from langchain_core.output_parsers import StrOutputParser
output_parser=StrOutputParser=StrOutputParser()
chain=prompt|llm|output_parser


response=chain.invoke({"input":"Can you tell me about AI"})
print(response)
```
